```python
import torch
import torch.nn as nn
from tokenizers import Tokenizer
```

```python
from torch.nn.utils.rnn import pad_sequence
```

```python
tokenizer = Tokenizer.from_file("Custom_tokenizer.json")
```

```python
sent = {
    "Virat Kholi is the king of the cricket",
    "The man, The myth, The legend",
    "Run Machine",
    "The Greatest of all time",
    "Most Sweetest Person"
}
```

```python
padding_id = tokenizer.token_to_id("[PAD]")
```

```python
print(padding_id)
```

```python
seq = []

for s in sent:
  encoding = tokenizer.encode(s)
  seq.append(torch.tensor(encoding.ids))

  print(encoding.tokens)
  print(encoding.ids)
```

```python
pad_val = padding_id if padding_id is not None else 0
padding_seq = pad_sequence(seq, batch_first=True, padding_value=float(pad_val))
```

```python
print(padding_seq)
```

```python
#showing the token into number so that attention mask can only mask to the 1 which is true
attention_mask = (padding_seq != padding_id).int()
print(attention_mask)
```
