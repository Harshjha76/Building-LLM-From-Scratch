```python
sentences = [
    "Artificial intelligence is changing the world.",
    "Python is a popular programming language.",
    "The sun rises in the east.",
    "Machine learning helps computers learn from data.",
    "I love learning new technologies.",
    "A computer consists of hardware and software.",
    "Consistency is the key to success.",
    "Large language models understand patterns in text.",
    "Programming improves logical thinking.",
    "Data is the foundation of artificial intelligence."
]
```

```python
word = []

for sentence in sentences:
  #extend -> basically store in the same list
  word.extend(sentence.split())

  #set me koi bhi cheez repeat nhi hota hai
  vocab = sorted(set(word))

  print(vocab)
```

```python
print(len(vocab))
```

```python
#word to id -> encoding
word_to_id = {
    word :index
    for index, word in enumerate(vocab)
}

print(word_to_id)
```

```python
#id to word -> decoding
id_to_word = {
    index :word
    for word,index in word_to_id.items()
}

print(id_to_word)
```

```python
tokenized_sentences = []

for sentence in sentences:
  tokens = sentence.split()
  token_id = [
      word_to_id[token]
      for token in tokens
  ]
  tokenized_sentences.append(token_id)
```

```python
print("Tokenized sentences:")
print(tokenized_sentences)
```

```python
print(sentence)
print(tokenized_sentences[0])
```

```python
x = []
y = []

for sentence in tokenized_sentences:
  for i in range(len(sentence)-1):
    x.append(sentence[i])
    y.append(sentence[i+1])
```

```python
print(x)
print(y)
```

```python
#check data type of x and y
print(type(x))
print(type(y))
```

```python
#converting x and y into tensor
import torch
x = torch.tensor(x,dtype=torch.long)
y = torch.tensor(y,dtype=torch.long)
```

```python
print(type(x))
print(type(y))
```

```python
import torch.nn as nn
```

```python
class NaturalLanguageModel(nn.Module):
  def __init__(self, vocab_size, embedding_dim):
    # calling the parent's constructor
    super().__init__()

    # embedding layer
    self.embedding = nn.Embedding(vocab_size, embedding_dim)

    # hidden layer
    self.hidden = nn.Linear(embedding_dim, 16)

    # output layer
    self.output = nn.Linear(16, vocab_size)

    # reLU function -> -ve not accepted because range(0 to aage tk)
    # gives the non-linearity on the data and if -ve value are there
    # considered as a 0
    self.relu = nn.ReLU()

  def forward(self, x):
    x = self.embedding(x)
    x = self.hidden(x)
    x = self.relu(x)
    x = self.output(x)
    return x
```

```python
embedding_dim = 7
vocab_size = len(vocab)

# Creating the object
model = NaturalLanguageModel(vocab_size, embedding_dim)
print(model)
```

```python
#defining loss and optimizer

import torch.optim as optim
loss_function = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
```

```python

epoches = 500

for epoch in range(epoches):
  prediction = model(x)
  loss = loss_function(prediction, y)
  optimizer.zero_grad()
  loss.backward()
  optimizer.step()
  if epoch % 100 == 0:
    print(f"Epoch: {epoch}, Loss: {loss.item()}")
```

```python
def predict_next_word(input_word):
  input_word = input_word.lower()
  if input_word not in word_to_id:
    print("Invalid input")

  input_id = word_to_id[input_word]
  input_tensor = torch.tensor([input_id], dtype=torch.long)
  prediction = model(input_tensor)

  model.eval()
  with torch.no_grad():
      prediction = model(input_tensor)
      predicted_id = torch.argmax(prediction).item()
  return id_to_word[predicted_id]
```

```python
print("\nNext Word Predictions:")

print("Artificial ->", predict_next_word("Artificial"))
```

```python
def generate_sentence(start_word, number_of_words=5):

    current_word = start_word.lower()

    if current_word not in word_to_id:
        return "Starting word not found"

    result = [current_word]

    model.eval()

    with torch.no_grad():

        for _ in range(number_of_words - 1):

            input_id = word_to_id[current_word]

            input_tensor = torch.tensor([input_id])

            output = model(input_tensor)

            predicted_id = torch.argmax(
                output,
                dim=1
            ).item()

            current_word = id_to_word[predicted_id]

            result.append(current_word)

    return " ".join(result)
```

```python
print("\nGenerated Sentences:")

print(generate_sentence("Artificial", 10))
```
