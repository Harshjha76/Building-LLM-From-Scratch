```python
model = "google/flan-t5-small"
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained(model)
model = AutoModelForSeq2SeqLM.from_pretrained(model)
```

```python
context = [
    "Artificial Intelligence is changing the world.",
    "AI enables machines to perform tasks that require human intelligence.",
    "Machine learning is a branch of artificial intelligence.",
    "AI systems learn patterns from large amounts of data.",
    "Chatbots can communicate with humans using natural language.",
    "Deep learning uses neural networks to solve complex problems.",
    "AI is widely used in healthcare, education, and finance.",
    "Self-driving cars use AI to understand their surroundings.",
    "Artificial intelligence can improve productivity and save time.",
    "Responsible AI development is important for the future."
]
```

```python

model.eval()
```

```python
question = input("Ask me question -")
```

```python
prompt = f'''
      Answer the questions only using the data in context.
      IF information is not available in context
      say : Information is not provided

      context : {context}
      question : {question}
      answer :
'''
```

```python
inputs = tokenizer(
    prompt,
    return_tensors="pt",
    max_length=512,
    truncation=True
)
```

```python
#generate answer
with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens = 100,
        num_beams = 4
    )
```

```python
# decode answer
answer = tokenizer.decode(outputs[0], skip_special_tokens=True)
```
