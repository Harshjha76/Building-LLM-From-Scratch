```python
model = "mrm8488/distilroberta-finetuned-financial-news-sentiment-analysis"
```

```python
from transformers import pipeline
```

```python
classifier = pipeline(
    model = model
)
```

```python
text = "Financial planning helps individuals manage their income, control expenses, save money, and achieve long-term financial stability."
```

```python
result = classifier(text)
print(text)
print(result[0]['label'])
print(result[0]["score"])
```
