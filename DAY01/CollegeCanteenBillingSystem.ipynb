```python
print("======= WELCOME TO COLLEGE CANTEEN =====\n")

print("1. Sandwich - Rs. 50","\n2. Burger - Rs. 50", "\n3. Cold Coffee -  Rs. 60","\n4. French Fries - Rs. 40 \n")

sand_quan = int(input("Enter quantity of sandwich: "))
burger_quan = int(input("Enter quantity of Burger: "))
coffee_quan = int(input("Enter quantity of Cold Coffee: "))
fries_quan = int(input("Enter quantity of French Fries: "))
print()

print("====== YOUR BILL ======\n")

total_sandcost = sand_quan * 50
total_burgercost = burger_quan * 50
total_coffeecost = coffee_quan * 60
total_friescost = fries_quan * 40

print(f"Sandwich : Rs. {total_sandcost}")
print(f"Burger : Rs. {total_burgercost}")
print(f"Cold Coffee : Rs. {total_coffeecost}")
print(f"French Fries : Rs. {total_friescost}","\n")

total_cost = total_sandcost + total_burgercost + total_coffeecost + total_friescost
print(f"Total Bill: Rs. {total_cost}")

if(total_cost > 500):
  discount = (total_cost/100)*20
  print(f"Discount : Rs. {discount}")
elif(total_cost >=300 and total_cost <=500):
  discount = (total_cost/100)*10
  print(f"Discount : Rs. {discount}")
else:
    discount = 0

final_Amount = total_cost - discount
print(f"Final Amount : Rs. {final_Amount} \n")

if(final_Amount > 400):
  print("Thank you for being our Premium Customer!")
else:
  print("Thank you! Visit Again!")
print("=====================================")
```
