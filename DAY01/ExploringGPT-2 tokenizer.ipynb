```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
```

```python
words = {
    "happiness",
    "programming",
    "playing",
    "artificialintelligence"
}
```

```python
for word in words:
  tokens = tokenizer.tokenize(word)
  print("Word :", word)
  print("Token :",tokens)
  print("Token IDs :",tokenizer.encode(tokens))
  print("---"*10)
```
