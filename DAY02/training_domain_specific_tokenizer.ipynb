```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace
```

```python
#if unknown words are there so tells unknown not showing error
tokenizer = Tokenizer(BPE(unk_token="[UNK]"))
```

```python
#first token will be whitespaces means spaces
tokenizer.pre_tokenizer = Whitespace()
```

```python
trainer = BpeTrainer(
    vocab_size = 5000,
    min_frequency = 2,
    special_tokens = ["[UNK]","[PAD]","[BOS]","[EOS]"],
)
```

```python
tokenizer.train(["/content/data.txt"],trainer=trainer)
```

```python
tokenizer.save("tokenizer.json")
```

```python
#finding the vocab size
vocab = tokenizer.get_vocab_size()
print(vocab)
```

```python
#creating one list
sent = [
    "artificial intelligence",
    "machine learning",
    "neural network",
    "mango"
]

for s in sent:
  output = tokenizer.encode(s)
  #printing whole word
  print(s)
  #printig tokens
  print(output.tokens)
  #printing token ids
  print(output.ids)
  print("------------------")
```
