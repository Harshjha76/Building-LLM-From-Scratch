```python
import torch
import torch.nn as nn
```

```python
#creating dictionary
vocab = {
    "machine" : 0,
    "learning" : 1,
    "python" : 2,
    "i" : 3,
    "like" : 4,
    "artifical" : 5,
    "neural" : 10,
    "network" : 13,
    "java" : 9,
    "deep" : 8,
    "and" : 87
}
```

```python
vocab_size = len(vocab)
print(vocab_size)
```

```python
embedding_dim = 7
```

```python
#using only those word which is present in the dictionary
sent = 'i like python java and machine learning'
```

```python
tokens = sent.split()
print(tokens)
```

```python
input_ids = [vocab[word] for word in tokens]
print(input_ids)
```

```python
print(type(input_ids))
```

```python
input_ids = torch.tensor(input_ids)
print(input_ids)
print(type(input_ids))
```

```python
embedding_layer = nn.Embedding(vocab_size, embedding_dim)
```

```python
token_embedding = embedding_layer(input_ids)

print(tokens)
print(input_ids)
print(token_embedding)
```
