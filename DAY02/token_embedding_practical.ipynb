```python
import torch
import torch.nn as nn
from tokenizers import Tokenizer
```

```python
tokenizer = Tokenizer.from_file("/content/tokenizer.json")
```

```python
vocab_size = tokenizer.get_vocab_size()
print(vocab_size)
```

```python
sent_1 = " machine learning is a subset of artificial intelligence"
```

```python
encoding = tokenizer.encode(sent_1)
input_ids = encoding.ids
```

```python
print(encoding.tokens)
print(input_ids)
```

```python
embedding_dim = 4
```

```python
token_embedding_layer = nn.Embedding(vocab_size, embedding_dim)
```

```python
print(type(input_ids))
```

```python
input_tensor = torch.tensor(input_ids)
```

```python
token_embedding = token_embedding_layer(input_tensor)
print(token_embedding)
```
