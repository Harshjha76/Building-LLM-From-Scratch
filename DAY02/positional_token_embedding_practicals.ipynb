```python
from tokenizers import Tokenizer
import torch
import torch.nn as nn

# 1. Load your trained tokenizer
tokenizer = Tokenizer.from_file("tokenizer.json")

# 2. Give a sentence
sentence = "machine learning is a subset of artificial intelligence"

# 3. Convert sentence into token IDs
encoding = tokenizer.encode(sentence)
input_ids = encoding.ids

print("Tokens:", encoding.tokens)
print("Token IDs:", input_ids)

# 4. Create token embedding layer
vocab_size = tokenizer.get_vocab_size()
embedding_dim = 4

token_embedding = nn.Embedding(vocab_size, embedding_dim)

# 5. Generate token embeddings
input_tensor = torch.tensor([input_ids])
token_embedding = token_embedding(input_tensor)
print("Token Embeddings:")
print(token_embedding)
```

```python
max_len = 30;
```

```python
positional_embedding_layer = nn.Embedding(max_len , embedding_dim)
```

```python
sequence_length = input_tensor.shape[1]
position_ids = torch.arange(sequence_length)
```

```python
print(position_ids)
```

```python
position_embedding = positional_embedding_layer(position_ids)
```

```python
print(position_embedding)
```

```python
final_embedding = token_embedding + position_embedding
print(final_embedding)
```
