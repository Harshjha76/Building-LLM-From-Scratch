```python
import requests
```

```python
from bs4 import BeautifulSoup
```

```python
urls = ["https://www.databricks.com/blog/machine-learning"]
data=[]
for url in urls:
  response = requests.get(url)
  soup = BeautifulSoup(response.text, "html.parser")
  for paragraphs in soup.find_all("p"):
    data.append(paragraphs.text)

print(len(data))
print(data)
```

```python
#put data in txt file
with open("data_02.txt","w") as f:
  for item in data:
    f.write("%s\n"% item)
```
